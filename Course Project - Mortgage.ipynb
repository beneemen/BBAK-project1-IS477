```python
#imports
import pandas as pd
```

```python
df = pd.read_csv('MORTGAGE30US.csv')

df
```

```python
df["observation_date"] = pd.to_datetime(df["observation_date"])

monthly_df = (
    df.groupby(df["observation_date"].dt.to_period("M"))["MORTGAGE30US"]
      .mean()
      .reset_index()
)

monthly_df["observation_date"] = monthly_df["observation_date"].dt.strftime("%b %Y")
monthly_df = monthly_df.rename(columns={
    "observation_date": "Month",
    "MORTGAGE30US": "Monthly Average"
})

monthly_df
```
